## **Bar charts**

**Draw a Bar Chart to the number of passengers who 'Survived'. Remember Survived with value 0 means did not survive, with value 1 means survived. Give you comments on the chart here:**

1.

```python
# Choose the relevant colum and Fill in the missing code

type_counts = df['Survived'].value_counts()

# Create the bar chart
plt.figure(figsize=(10, 6))
type_counts.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title("number of passengers who 'Survived")
plt.xlabel('Survived')
plt.ylabel('passengers count')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

# **Exercise 3. Comparing Data for relationships, correlations and trends**

**Draw a scatter plot to check the correlation between 'Age' and the 'Fare', Write your comments here:**

```python
# Choose the relevant columns and fill in the missing code

df.columns
# Create the scatter plot with regression line
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Age', y='Fare', color='teal', edgecolor='black', alpha=0.7)
sns.regplot(data=df, x='Age', y='Fare', scatter=False, color='darkred', line_kws={'label':"Linear Fit"})

plt.title("correlation between 'Age' and the 'Fare'")
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
